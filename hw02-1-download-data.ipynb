```python
import os
import boto3

SOURCE_BUCKET = "dsan6000-wikipedia"
SOURCE_PREFIX = "hourly_parquet/"
LOCAL_DIR = "data"

os.makedirs(LOCAL_DIR, exist_ok=True)
s3 = boto3.client("s3", region_name="us-east-1")
```

```python
response = s3.list_objects_v2(Bucket=SOURCE_BUCKET, Prefix=SOURCE_PREFIX)
parquet_keys = [obj["Key"] for obj in response["Contents"] if obj["Key"].endswith(".parquet")]
print(f"Found {len(parquet_keys)} parquet files")
parquet_keys[:3]
```

```python
for key in parquet_keys:
    filename = os.path.basename(key)
    s3.download_file(SOURCE_BUCKET, key, os.path.join(LOCAL_DIR, filename))
    print(f"Downloaded {filename}")
```

```python
local_files = sorted(f for f in os.listdir(LOCAL_DIR) if f.endswith(".parquet"))
print(f"{len(local_files)} parquet files in {LOCAL_DIR}/")
local_files
```
