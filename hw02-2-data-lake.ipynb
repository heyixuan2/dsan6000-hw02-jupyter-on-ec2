```python
import os
import boto3

BUCKET_NAME = "dsan6000-yh973"
DEST_PREFIX = "wikipedia-hourly/"
LOCAL_DIR = "data"

s3 = boto3.client("s3", region_name="us-east-1")
```

```python
try:
    s3.create_bucket(Bucket=BUCKET_NAME)
    print(f"Created bucket s3://{BUCKET_NAME}")
except s3.exceptions.BucketAlreadyOwnedByYou:
    print(f"Bucket s3://{BUCKET_NAME} already exists (owned by you)")
```

```python
pab = s3.get_public_access_block(Bucket=BUCKET_NAME)
pab["PublicAccessBlockConfiguration"]
```

```python
local_files = sorted(f for f in os.listdir(LOCAL_DIR) if f.endswith(".parquet"))
for filename in local_files:
    s3.upload_file(os.path.join(LOCAL_DIR, filename), BUCKET_NAME, DEST_PREFIX + filename)
    print(f"Uploaded {filename} -> s3://{BUCKET_NAME}/{DEST_PREFIX}{filename}")
```

```python
response = s3.list_objects_v2(Bucket=BUCKET_NAME, Prefix=DEST_PREFIX)
uploaded = [obj["Key"] for obj in response["Contents"]]
print(f"{len(uploaded)} objects in s3://{BUCKET_NAME}/{DEST_PREFIX}")
uploaded[:3]
```
